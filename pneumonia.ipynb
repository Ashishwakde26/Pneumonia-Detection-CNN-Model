```python
import os
import cv2
import numpy as np
```

```python
labels = ['PNEUMONIA', 'NORMAL']
img_resize = 150

def get_data(dir):
    data = []
    for lable in labels:
        path = os.path.join(dir, lable)
        class_num = labels.index(lable)

        for img in os.listdir(path):
            try:
                if img.endswith('jpeg'):
                    img_arr = cv2.imread(os.path.join(path,img), cv2.IMREAD_GRAYSCALE)
                    resize_arr = cv2.resize(img_arr, (img_resize, img_resize))
                    data.append([resize_arr, class_num])
            except Exception as e:
                print(e)
    return np.array(data,dtype= 'object')
```

```python
train = get_data('chest_xray/train')
```

```python
test = get_data('chest_xray/test')
val = get_data('chest_xray/val')
```
